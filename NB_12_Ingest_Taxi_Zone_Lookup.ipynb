```python
import pandas as pd
import requests

# Official NYC TLC Taxi Zone Lookup CSV
url = "https://d37ci6vzurychx.cloudfront.net/misc/taxi_zone_lookup.csv"

response = requests.get(url, timeout=60)
response.raise_for_status()

# Read CSV
from io import BytesIO

zone_pd = pd.read_csv(BytesIO(response.content))

print("Downloaded records:", len(zone_pd))
print("Columns:", zone_pd.columns.tolist())

# Standardize column names
zone_pd.columns = [
    "location_id",
    "borough",
    "zone_name",
    "service_zone"
]

# Convert to Spark DataFrame
zone_df = spark.createDataFrame(zone_pd)

# Save in Bronze Lakehouse
(
    zone_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("taxi_zone_lookup_raw")
)

print("Taxi Zone Lookup saved successfully!")
print("Total records:", zone_df.count())

display(zone_df.limit(10))
```
