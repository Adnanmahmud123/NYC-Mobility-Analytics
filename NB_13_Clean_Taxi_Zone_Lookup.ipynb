```python
from pyspark.sql import functions as F

# Read official zone lookup from Bronze
zone_df = spark.read.table(
    "NYC_Mobility_Analytics.NYC_Bronze_LH.dbo.taxi_zone_lookup_raw"
)

print("Original records:", zone_df.count())

# Clean and standardize
clean_zone_df = (
    zone_df
    .select(
        F.col("location_id").cast("int").alias("location_id"),
        F.initcap(F.trim("borough")).alias("borough"),
        F.trim("zone_name").alias("zone_name"),
        F.trim("service_zone").alias("service_zone")
    )
    .filter(F.col("location_id").isNotNull())
    .filter(F.col("borough").isNotNull())
    .filter(F.col("zone_name").isNotNull())
    .dropDuplicates(["location_id"])
    .orderBy("location_id")
)

# Save to Silver Lakehouse
(
    clean_zone_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("taxi_zone_lookup_clean")
)

print("Taxi Zone Lookup Silver data saved successfully!")
print("Clean records:", clean_zone_df.count())

display(clean_zone_df.limit(20))
```
