```python
import requests
import pandas as pd

# World Bank GDP API - United States
url = (
    "https://api.worldbank.org/v2/country/USA/"
    "indicator/NY.GDP.MKTP.CD"
)

params = {
    "format": "json",
    "per_page": 1000
}

# Fetch GDP data
response = requests.get(url, params=params, timeout=60)
response.raise_for_status()

data = response.json()

if len(data) < 2 or data[1] is None:
    raise Exception("World Bank API returned no GDP data!")

records = []

for item in data[1]:
    if item["value"] is not None:
        records.append({
            "year": int(item["date"]),
            "country_code": "USA",
            "country_name": "United States",
            "indicator": "GDP (current US$)",
            "gdp_current_usd": float(item["value"])
        })

# Create Pandas DataFrame
gdp_pd = pd.DataFrame(records)

print("Downloaded GDP records:", len(gdp_pd))

# Convert to Spark DataFrame
gdp_df = spark.createDataFrame(gdp_pd)

# Save raw data to Bronze Lakehouse
(
    gdp_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("worldbank_gdp_raw")
)

print("World Bank GDP data saved successfully!")
print("Total records:", gdp_df.count())

display(gdp_df.orderBy("year"))
```
